```python
import pandas as pd
import numpy as np
print(f"Pandas Version: {pd.__version__}")
```

## ১. রিয়েলিস্টিক ই-কমার্স ডেটাসেট তৈরি ও পরিদর্শন

```python
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'customer_id': range(1001, 1001 + n),
    'city': np.random.choice(['Dhaka', 'Chittagong', 'Sylhet', 'Rajshahi'], size=n, p=[0.5, 0.25, 0.15, 0.1]),
    'age': np.random.choice([20, 25, 30, 35, 45, 60, np.nan], size=n),
    'annual_spend': np.random.uniform(5000, 150000, size=n),
    'is_churn': np.random.choice([0, 1], size=n, p=[0.7, 0.3])
})
print("Shape:", df.shape)
df.head()
```

## ২. মিসিং ডেটা হ্যান্ডলিং ও ইম্পিউটেশন

```python
# মিসিং ভ্যালু চিহ্নিত করা
print("Missing Count:\n", df.isna().sum())

# শহরের গড় বয়স দিয়ে শূন্যস্থান পূরণ (Group Imputation)
df['age'] = df.groupby('city')['age'].transform(lambda x: x.fillna(x.median()))
print("\nইম্পিউটেশনের পর মিসিং কাউন্ট:", df['age'].isna().sum())
```

## ৩. গ্রুপ অপারেশন ও ফিচার ইঞ্জিনিয়ারিং (Transform)

```python
# ১. প্রতিটি শহরের গড় বাৎসরিক খরচ বের করা
df['city_avg_spend'] = df.groupby('city')['annual_spend'].transform('mean')

# ২. খরচের অনুপাত (Spend Ratio Feature)
df['spend_ratio_to_city'] = df['annual_spend'] / df['city_avg_spend']

# ৩. বাকেটিং / ক্যাটাগরিকাল বিন তৈরি (pd.qcut)
df['spend_tier'] = pd.qcut(df['annual_spend'], q=3, labels=['Low', 'Medium', 'High'])

df[['city', 'annual_spend', 'city_avg_spend', 'spend_ratio_to_city', 'spend_tier']].head()
```

## ৪. ডেটা টাইপ অপটিমাইজেশন (Memory Saving)

```python
mem_before = df.memory_usage(deep=True).sum()
df['city'] = df['city'].astype('category')
df['is_churn'] = df['is_churn'].astype('int8')
mem_after = df.memory_usage(deep=True).sum()
print(f"মেমরি সাশ্রয় হয়েছে: {(mem_before - mem_after) / mem_before * 100:.2f}%")
```
