## ১. অ্যারে তৈরি ও মেমরি পরীক্ষা

```python
# ১-ডি ভেক্টর ও ২-ডি ম্যাট্রিক্স
vec = np.array([1, 2, 3, 4, 5], dtype=np.float32)
mat = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)
eye_mat = np.eye(3)

print("Vector shape:", vec.shape)
print("Matrix shape:", mat.shape)
print("Identity matrix:\n", eye_mat)
```

## ২. ব্রডকাস্টিং নীতি ও ডেটা নরমালাইজেশন

```python
# ML-এ ফিচার স্ট্যান্ডার্ডাইজেশন (Standardization) ব্রডকাস্টিং দিয়ে
np.random.seed(42)
X = np.random.randint(10, 100, size=(5, 3)).astype(float)
print("আসল ডেটা X:\n", X)

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

# ব্রডকাস্টিং প্রয়োগ: (5, 3) থেকে (3,) বিয়োগ ও ভাগ
X_scaled = (X - mean) / std
print("\nস্ট্যান্ডার্ডাইজড ডেটা X_scaled (গড়=০, ভ্যারিয়্যান্স=১):\n", np.round(X_scaled, 3))
print("নতুন গড়:", np.round(np.mean(X_scaled, axis=0), 2))
```

## ৩. রৈখিক বীজগণিত ও স্ক্র্যাচ লিনিয়ার রিগ্রেশন

```python
# Normal Equation: theta = (X^T X)^-1 X^T y
# সত্য সমীকরণ: y = 2*x1 + 3*x2 + 5
n_samples = 100
x1 = 2 * np.random.rand(n_samples, 1)
x2 = 3 * np.random.rand(n_samples, 1)
y = 5 + 2*x1 + 3*x2 + np.random.randn(n_samples, 1) * 0.1

# বায়াস কলাম 1 যোগ করা
X_b = np.c_[np.ones((n_samples, 1)), x1, x2]

# OLS সমাধান
theta_best = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print("Estimated Parameters [Bias, W1, W2]:\n", theta_best.ravel())
```

## ৪. ভেক্টরাইজেশন বনাম লুপ গতির তুলনা

```python
N = 5_000_000
a = np.arange(N, dtype=np.float64)
b = np.arange(N, dtype=np.float64)

# NumPy ভেক্টরাইজেশন
t0 = time.time()
c_np = a * b + 2.0
t_np = time.time() - t0
print(f"NumPy সময়: {t_np:.5f} সেকেন্ড")
```
