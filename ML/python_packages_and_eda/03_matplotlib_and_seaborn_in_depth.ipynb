```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

sns.set_theme(style="whitegrid")
print("Libraries loaded successfully!")
```

## ১. লস কনভার্জেন্স কার্ভ (Loss Curve) তৈরি

```python
# কৃত্রিম লস ডেটা তৈরি
epochs = np.arange(1, 31)
train_loss = 2.5 * np.exp(-epochs / 6) + 0.1 * np.random.randn(30) * 0.2
val_loss = 2.5 * np.exp(-epochs / 6) + 0.3 + 0.02 * (epochs / 5)**2

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(epochs, train_loss, 'b-o', label='Training Loss', markersize=4)
ax.plot(epochs, val_loss, 'r-s', label='Validation Loss (Overfitting)', markersize=4)
ax.axvline(x=15, color='gray', linestyle='--', label='Early Stopping Point')

ax.set_title("Model Training vs Validation Loss Curve", fontsize=12, fontweight='bold')
ax.set_xlabel("Epochs")
ax.set_ylabel("Cross-Entropy Loss")
ax.legend()
plt.show()
```

## ২. কোরিলেশন হিটম্যাপ (Feature Correlation Heatmap)

```python
np.random.seed(42)
df = pd.DataFrame(np.random.randn(100, 5), columns=['Age', 'Income', 'Credit_Score', 'Debt', 'Default'])
df['Debt'] = df['Income'] * 0.8 + np.random.randn(100) * 0.2 # স্ট্রং কোরিলেশন যুক্ত করা

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(df.corr(), annot=True, cmap='vlag', fmt='.2f', vmin=-1, vmax=1, ax=ax)
ax.set_title("Multicollinearity Check via Correlation Matrix")
plt.show()
```

## ৩. কনফিউশন ম্যাট্রিক্স হিটম্যাপ

```python
from sklearn.metrics import confusion_matrix

y_true = np.array([0, 0, 0, 1, 1, 1, 0, 1, 0, 1])
y_pred = np.array([0, 0, 1, 1, 1, 0, 0, 1, 0, 1])

cm = confusion_matrix(y_true, y_pred)
fig, ax = plt.subplots(figsize=(4, 3))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Non-Fraud', 'Fraud'], yticklabels=['Non-Fraud', 'Fraud'], ax=ax)
ax.set_xlabel('Predicted')
ax.set_ylabel('Actual')
ax.set_title('Production Model Confusion Matrix')
plt.show()
```
